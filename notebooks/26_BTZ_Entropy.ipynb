# Notebook 26: BTZ Entropy from Palindromic Wall-Crossing (Paper II, §10.8)

The BTZ entropy decomposes into discrete (wall-crossing) and continuous (one-loop) parts:
$$S_{\text{BTZ}}(N) = S_{\text{1-loop}} + \Delta S_{\text{wc}}$$

Even $N$: $\Delta S = \log 2$ (1 bit, self-paired $m = N/2$).
Odd $N$: $\Delta S = 2\log 2$ (2 bits, paired modes $m, N{-}m$).
Through $N = 12$: exactly 9 bits.

```python
import sys, math
sys.path.insert(0, '../src')
from planetary_polygons.extensions.btz_entropy import (
    wall_crossing_entropy, wall_crossing_table, one_loop_entropy,
    entropy_decomposition, morse_index_L2, fractional_morse_index,
    find_threshold, palindromic_field, BOLZA_EIGENVALUES_TRIVIAL
)
```

## 1. Wall-crossing table: 1 or 2 bits per threshold

```python
rows = wall_crossing_table(N_max=14)
print(f"{'N':>4} {'type':>12} {'\u0394S':>10} {'field':>18} {'cum bits':>10}")
print('-'*58)
for N, dS, sp, field, cum in rows:
    tp = 'self-paired' if sp else 'paired'
    print(f"{N:4d} {tp:>12} {dS/math.log(2):10.0f} bit{'s' if not sp else ' '} {field:>18} {cum:10.1f}")
print(f'\nThrough N=12: exactly {rows[5][4]:.0f} bits.')
```

## 2. Entropy decomposition: discrete vs continuous

```python
result = entropy_decomposition(N_min=7, N_max=17)
print(f"{'N':>4} {'S_1loop':>10} {'S_wc':>8} {'cum_1loop':>10} {'cum_wc':>8} {'total':>10} {'bits':>8}")
print('-'*60)
S1, Sw = 0, 0
for t in result['terms']:
    S1 += t['S_1loop']; Sw += t['S_wc']
    print(f"{t['N']:4d} {t['S_1loop']:10.2f} {t['S_wc']:8.2f} {S1:10.2f} {Sw:8.2f} {S1+Sw:10.2f} {(S1+Sw)/math.log(2):8.1f}")
print(f"\nDiscrete: {result['bits_wc']:.0f} bits ({result['S_wc_total']/result['S_total']*100:.1f}%)")
print(f"Continuous: {(result['S_1loop_total'])/math.log(2):.0f} bits ({result['S_1loop_total']/result['S_total']*100:.1f}%)")
```

## 3. L²-Morse index on the Bolza surface

```python
print("Fractional Morse index \u03c4(P\u208b) at palindromic thresholds:\n")
print(f"{'N':>4} {'\u03c1*':>10} {'\u03c4':>12} {'n_neg':>6} {'frac(\u03c4)':>12}")
print('-'*48)
for N in range(5, 14):
    rho = find_threshold(N)
    if rho and rho > 0.01:
        tau, n_neg, closest = fractional_morse_index(N, rho)
        frac = tau - int(tau)
        print(f"{N:4d} {rho:10.4f} {tau:12.6f} {n_neg:6d} {frac:12.6f}")
print('\nThe denominator is 2 (half-integer from the eta invariant).\nConsistent with K-theoretic framework.')
```

## 4. One-loop entropy growth

```python
print(f"{'N':>4} {'S_1loop':>12} {'S/log(N)':>12} {'~(N/2)logN?':>12}")
print('-'*44)
for N in range(6, 18):
    S = one_loop_entropy(N)
    approx = N/2 * math.log(N)
    print(f"{N:4d} {S:12.4f} {S/math.log(N) if N>1 else 0:12.4f} {approx:12.4f}")
print('\nS_1loop ~ (N/2)log N from the double factorial of Casimir gaps.')
```

## Summary

1. **Wall-crossing**: even $N$ = 1 bit, odd $N$ = 2 bits. Through $N=12$: 9 bits.
2. **Decomposition**: discrete ~9%, continuous ~91% of total entropy.
3. **L²-Morse index**: half-integer at thresholds (eta invariant).
4. **One-loop**: $S \sim (N/2)\log N$ from the double factorial.
