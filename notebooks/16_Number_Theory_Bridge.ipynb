**Pythagorean sign rule:** For $c = m^2 + n^2$ prime with $c \equiv 1 \pmod{8}$:
$$a_c = 2\,(-1)^{e/4}\,(2/o)$$
where $e$ is the even and $o$ the odd Pythagorean parameter.

## 1. Bolza form $\eta(8z)\eta(16z)$: first coefficients

```python
print('Bolza form coefficients a_n for n ≤ 30:')
for n in range(1, 31):
    a = bolza_coefficient(n)
    if a != 0:
        print(f'  a_{n} = {a}')
```

## 2. Vanishing criterion: $a_p \ne 0$ iff $p \equiv 1 \pmod{8}$

```python
from planetary_polygons.extensions.number_theory.sieve import sieve
primes = sieve(1000)
violations = 0
for p in primes:
    if p < 3: continue
    a = bolza_coefficient(p)
    should_be_nonzero = (p % 8 == 1)
    is_nonzero = (a != 0)
    if should_be_nonzero != is_nonzero:
        violations += 1
        print(f'VIOLATION at p={p}: a_p={a}, p mod 8 = {p%8}')
print(f'Tested {len([p for p in primes if p >= 3])} primes, {violations} violations.')
```

## 3. Pythagorean sign rule verification

```python
count = 0; ok = 0
for p in primes:
    if p < 3 or p % 8 != 1: continue
    a = bolza_coefficient(p)
    # Find Pythagorean decomposition p = m² + n²
    found = False
    for m in range(1, int(p**0.5)+1):
        n2 = p - m*m
        if n2 <= 0: continue
        n = int(round(n2**0.5))
        if n*n == n2 and m > n > 0 and (m-n) % 2 == 1:
            e, o = (m, n) if m % 2 == 0 else (n, m)
            sign = 2 * ((-1)**(e//4)) * (1 if o % 8 in [1,7] else -1)
            if sign == a: ok += 1
            else: print(f'FAIL: p={p}, a={a}, predicted={sign}')
            count += 1; found = True; break
    if not found and p % 8 == 1:
        print(f'No decomposition found for p={p}')
print(f'\nVerified: {ok}/{count} primes match the Pythagorean sign rule.')
```
