## Mass Predictions

```python
table = mass_table()
print(f"{'Quark':>6} {'Prediction':>12} {'Observed':>12} {'Unit':>5}")
print("-" * 38)
for q in ['t', 'b', 'c', 'u']:
    d = table[q]
    val = d['pred']
    obs = d['obs']
    if val < 0.01:
        print(f"{q:>6} {val*1e3:12.3f} {obs*1e3:12.3f} {'MeV':>5}")
    else:
        print(f"{q:>6} {val:12.3f} {obs:12.3f} {'GeV':>5}")
```
