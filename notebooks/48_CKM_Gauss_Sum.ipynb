## The Gauss Sum Chain

```python
import numpy as np, cmath
N = 7
omega = np.exp(2j * pi / N)

# Quadratic residues
QR = sorted(set(a**2 % N for a in range(1, N)))
print(f"QR(7) = {QR}")

# Gauss sum
G_QR = sum(omega**a for a in QR)
alpha_2 = 1 + G_QR
print(f"G(QR) = {G_QR:.6f}")
print(f"Expected (-1+i*sqrt(7))/2 = {(-1+1j*sqrt(7))/2:.6f}")
print(f"alpha_2 = 1 + G(QR) = {alpha_2:.6f}")
print(f"arg(alpha_2) = {degrees(cmath.phase(alpha_2)):.4f} deg")
print(f"arctan(sqrt(7)) = {degrees(atan(sqrt(7))):.4f} deg")
print(f"sin^2(delta) = {sin(atan(sqrt(7)))**2:.6f} = 7/8 = {7/8}")
```
