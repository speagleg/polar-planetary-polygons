# 39. The Master Equation

From a=4 (spectral diameter) and Onsager selection of the icosahedron,
derive all E₈ data: b=a+1, c=a+2, |I*|=120, h=30, rank=8, dim=248.

```python
from fractions import Fraction
import math
```

## Step 1: b = a+1 from the Schwarz spherical bound

```python
a = 4  # spectral diameter (Perron-Frobenius, universal for bipartite ADE)

# E-type family: (p,q) = (2,3), unique non-dihedral pair
# (p-1)(q-1) = 2 has unique solution p=2, q=3
for p in range(2, 10):
    for q in range(p, 10):
        if (p-1)*(q-1) == 2:
            print(f"(p-1)(q-1) = 2: unique solution (p,q) = ({p},{q})")

p, q = 2, 3
# Spherical condition: 1/p + 1/q + 1/r > 1 → r < pq/(pq-p-q) = 6
r_crit = Fraction(p*q, p*q - p - q)
print(f"r_crit = pq/(pq-p-q) = {r_crit} = a+2 = {a+2}")
r_max = int(r_crit) - 1
print(f"r_max = {r_max} = a+1 = {a+1}")
b = r_max
print(f"b = {b}")
```

## Step 2: c = a+2 from the affine mark recurrence

```python
# On the longest arm of the affine E₈ diagram:
# d_{k+1} = 2d_k - d_{k-1}, with d_0=1, d_1=2
r = 5  # arm length = b = a+1
d = [1, 2]
for k in range(1, r):
    d.append(2*d[-1] - d[-2])
print(f"Marks on longest arm: {d}")
c = d[-1]
print(f"c = d_max = {c} = a+2 = {a+2}")
```

## Steps 3-7: All E₈ data

```python
# Step 3: |I*| = abc (Gauss-Bonnet)
I_star = a * b * c
epsilon = Fraction(1,2) + Fraction(1,3) + Fraction(1,5) - 1
print(f"|I*| = abc = {a}×{b}×{c} = {I_star}")
print(f"Check: a/ε = {a}/{epsilon} = {Fraction(a,1)/epsilon}")
print()

# Step 4: h = |I*|/a (mark-balance k₁=a)
h = I_star // a
print(f"h = |I*|/a = {I_star}/{a} = {h}")
print()

# Step 5: rank = 2a (class equation of I*)
# |CC(I*)| = 2b - 1 = 9, rank = 9 - 1 = 8
cc = 2*b - 1
rank = cc - 1
print(f"|CC(I*)| = 2b-1 = {cc}, rank = {rank} = 2a = {2*a}")
print()

# Step 6: Σ pivots = a + a/2 + a/4 = 7
pivots_sum = a + a//2 + a//4
print(f"Σ pivots = {a} + {a//2} + {a//4} = {pivots_sum} = N_crit")
print()

# Step 7: dim(E₈) = |I*| + 2^(rank-1)
dim_e8 = I_star + 2**(rank-1)
print(f"dim(E₈) = {I_star} + {2**(rank-1)} = {dim_e8}")
print(f"         = |I*| + 2^(rank-1) = SO(16) adj + spinor")
```
