# 45. BF Threshold μ₄=1 as Dirac Ground State

The first massive fermion level on H², not a choice.

## The Dirac spectrum on H²

Mass-dimension relation: Δ = |μ| + 1/2 (spin contribution)
- Δ = 1/2: massless (|μ| = 0)
- Δ = 1: BF boundary (|μ| = 1/2, massless fermion)
- Δ = 3/2: first massive level (|μ| = 1 = μ₄)

```python
print("Dirac spectrum on H²:")
print(f"  Δ = |μ| + 1/2  (the 1/2 is the spin contribution)")
print()
for mu in [0, 0.5, 1, 1.5, 2, 2.5]:
    Delta = mu + 0.5
    M2 = Delta * (Delta - 1)
    status = ""
    if mu == 0: status = "← vacuum"
    elif mu == 0.5: status = "← massless fermion (BF boundary)"
    elif mu == 1: status = "← μ₄ = 1: GROUND STATE of massive sector"
    elif mu > 1: status = f"← decouples below M_KK"
    print(f"  |μ| = {mu:4.1f}  →  Δ = {Delta:4.1f}  →  M² = {M2:5.2f}  {status}")

print()
print("μ₄ = 1 is the ground state: the lightest massive fermion on H².")
print("Higher levels (|μ| = 2, 3, ...) are heavier by M_KK and decouple.")
```
