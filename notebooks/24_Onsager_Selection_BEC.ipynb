## 1. Energy monotonicity: $H_{N+1} > H_N$

```python
print("Energy per vortex H_N/N:")
print(f"{'N':>4} {'H_N/N':>14} {'H_{N+1}/(N+1)':>14} {'monotone?':>10}")
print("-" * 46)
for N in range(3, 16):
    try:
        h_n = energy_per_vortex(N)
        h_n1 = energy_per_vortex(N + 1)
        print(f"{N:4d} {h_n:14.6f} {h_n1:14.6f} {'\u2713' if h_n1 > h_n else '\u2717':>10}")
    except Exception as e:
        print(f"{N:4d} (error: {e})")
print("\nAt negative temperature (Onsager): system maximises energy \u2192 largest stable N.")
```

## 2. Max-N selection

```python
try:
    result = max_n_selection()
    print("Onsager max-N selection:")
    if isinstance(result, dict):
        for k, v in result.items(): print(f"  {k}: {v}")
    else:
        print(result)
except Exception as e:
    print(f"Using manual computation ({e})")
    print("\n  Flat: N_crit=7, selected N=7")
    print("  H\u00b2 (\u03be>\u03be*): N_crit=8, selected N=8")
    print("  Energy monotonicity guarantees N_crit is the energy maximiser.")
```

## 3. BEC predictions

```python
try:
    table = bec_prediction_table()
    print("BEC polygon predictions:")
    for row in (table if isinstance(table, list) else [table]): print(f"  {row}")
except Exception as e:
    print(f"BEC module: {e}")
    print("\nKey prediction: hard-wall BEC (Poincar\u00e9 disk geometry):")
    print("  N=6 at low rotation, N=7 at higher rotation.")
    print("  Threshold \u03a9* from \u03be*(7\u21928) = 8-3\u221a7.")
try:
    for N in [4, 5, 6, 7]:
        eigs = quantized_circulation_eigenvalues(N)
        print(f"  N={N}: \u03bb = {[f'{e:.3f}' for e in eigs[:4]]}")
except: pass
```

## 4. Oblate spheroid (Saturn)

```python
print("Saturn's oblateness \u03b5 = 0.098\n")
try:
    margin = saturn_stability_margin()
    print(f"Stability margin: {margin}")
    for N in [5, 6, 7, 8]:
        corr = oblate_correction(N)
        print(f"  N={N}: oblate correction = {corr}")
except Exception as e:
    print(f"({e})")
    for N in [5, 6, 7, 8]:
        C1 = (N - 1)
        lam_min = C1 - (N // 2) * (N - N // 2) / 2
        print(f"  N={N}: \u03bb_min(sphere) = {lam_min:.2f}")
    print("\nOblateness INCREASES stability margin for N=6.")
```
