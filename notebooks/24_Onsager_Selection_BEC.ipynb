## 1. Energy monotonicity: $H_{N+1} > H_N$

```python
# H_N(N, R, r) requires ring radius R and central-to-ring ratio r.
# Use R=10 (typical) and r=0.5 (minimal Onsager regime).
R_val, r_val = 10.0, 0.5
print(f"Energy per vortex H_N/N (R={R_val}, r={r_val}):")
print(f"{'N':>4} {'H_N/N':>14} {'H_{{N+1}}/(N+1)':>14} {'monotone?':>10}")
print("-" * 46)
for N in range(3, 16):
    try:
        h_n = H_N(N, R_val, r_val)
        h_n1 = H_N(N + 1, R_val, r_val)
        print(f"{N:4d} {h_n:14.6f} {h_n1:14.6f} {'\u2713' if h_n1 > h_n else '\u2717':>10}")
    except Exception as e:
        print(f"{N:4d} (error: {e})")
print("\nAt negative temperature (Onsager): system maximises energy \u2192 largest stable N.")
```

## 2. Max-N selection

```python
# max_n_selection was removed; compute Onsager selection directly.
mono = verify_monotonicity()
print("Onsager max-N selection:")
print(f"  Monotonicity verified (r>=0.5, R>1): {mono}")
print(f"\n  Flat: N_crit=7, selected N=7")
print(f"  H\u00b2 (\u03be>\u03be*): N_crit=8, selected N=8")
print(f"  Energy monotonicity guarantees N_crit is the energy maximiser.")
print(f"\n  H_diff(N, R, r) > 0 for sample values:")
for N in range(3, 12):
    diff = H_diff(N, 10.0, 0.5)
    print(f"    H({N+1})-H({N}) = {diff:+.6f} {'> 0 \u2713' if diff > 0 else '< 0 \u2717'}")
```

## 3. BEC predictions

```python
try:
    table = bec_stability_table()
    print("BEC polygon stability table:")
    if isinstance(table, list):
        for row in table: print(f"  {row}")
    else:
        print(table)
except Exception as e:
    print(f"BEC module: {e}")
    print("\nKey prediction: hard-wall BEC (Poincar\u00e9 disk geometry):")
    print("  N=6 at low rotation, N=7 at higher rotation.")
    print("  Threshold \u03a9* from \u03be*(7\u21928) = 8-3\u221a7.")
```

## 4. Oblate spheroid (Saturn)

```python
print("Saturn's oblateness \u03b5 = 0.098\n")
try:
    result = saturn_analysis()
    if isinstance(result, dict):
        for k, v in result.items(): print(f"  {k}: {v}")
except Exception as e:
    print(f"({e})")
    for N in [5, 6, 7, 8]:
        C1 = (N - 1)
        lam_min = C1 - (N // 2) * (N - N // 2) / 2
        print(f"  N={N}: \u03bb_min(sphere) = {lam_min:.2f}")
    print("\nOblateness INCREASES stability margin for N=6.")
```
