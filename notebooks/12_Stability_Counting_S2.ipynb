# Notebook 12: Stability Counting on $S^2$ — $\mathcal{N} = 71/60$ (Paper I, §5)

$$\mathcal{N}(\Lambda) = \sum_{N=3}^{7}(1 - \cos\varphi_{\mathrm{crit}}(N)) = \frac{1}{2} + \frac{1}{3} + \frac{1}{4} + \frac{1}{10} + 0 = \frac{71}{60}$$

All S² thresholds are rational: $\cos\varphi_{\mathrm{crit}} = m(N{-}m)/(2(N{-}1))$.

```python
import sys
sys.path.insert(0, '../src')
from fractions import Fraction
from planetary_polygons.extensions.algebraic_thresholds import sphere_stability_threshold
```

## 1. Exact S² thresholds

```python
print('S² thresholds (all rational):')
print(f\"{'N':>3} {'ξ_crit':>10} {'cos(φ)':>10} {'1-cos(φ)':>10}\")
print('-' * 40)
total = Fraction(0)
for N in range(3, 8):
    m = N // 2
    cos_phi = Fraction(m*(N-m), 2*(N-1))
    one_minus = 1 - cos_phi
    total += one_minus
    xi = sphere_stability_threshold(N)
    xi_str = str(xi) if xi else '0'
    print(f'{N:3d} {xi_str:>10} {str(cos_phi):>10} {str(one_minus):>10}')
print(f'\nN(Λ) = {total} = {float(total):.10f}')
```

## 2. The $13^2/60$ identity

```python
sum_cos = Fraction(1,2)+Fraction(2,3)+Fraction(3,4)+Fraction(9,10)
print(f'Σ cos(φ_crit) = {sum_cos} = {sum_cos.numerator}/{sum_cos.denominator}')
print(f'{sum_cos.numerator} = 13² = {13**2}')
print(f'N(Λ) = 4 - 13²/60 = {4 - sum_cos} = 71/60')
```

## 3. Odd/even closed forms

```python
print('Odd N=2k+1: s = (3-k)/4')
for k in [1,2,3]:
    N = 2*k+1
    s = Fraction(3-k, 4)
    print(f'  N={N} (k={k}): s = {s} = {float(s):.4f}')
print('\nEven N=2k: s = (4k-2-k²)/(2(2k-1))')
for k in [2,3]:
    N = 2*k
    s = Fraction(4*k-2-k**2, 2*(2*k-1))
    print(f'  N={N} (k={k}): s = {s} = {float(s):.4f}')
```

3. Odd/even closed forms verified.
