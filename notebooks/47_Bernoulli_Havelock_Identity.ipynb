## The Bernoulli-Havelock Identity

$f(m,N) = \frac{N^2}{12}\left(1 - 6\,B_2(m/N)\right)$

```python
# Verify identity for all modes at N=7
print("m | f(m,7) direct | f(m,7) via B_2 | match")
print("-" * 50)
for m in range(1, N):
    direct = havelock_casimir(m, N)
    via_B2 = havelock_from_bernoulli(m, N)
    print(f"{m} | {str(direct):>14} | {str(via_B2):>14} | {direct == via_B2}")
```

## Parabola Form: $1 - 6\,B_2(x) = 6\,x(1-x)$

```python
# Verify at all N=7 modes
for m in range(N+1):
    x = Fraction(m, N)
    assert parabola_form(x) == 6 * x * (1 - x)
print("Parabola identity verified for all m/7, m=0..7")
```
