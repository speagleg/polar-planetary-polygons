# Notebook 13: Growth Law $\rho^* \sim N^2/24$ from $B_2 = 1/6$ (Paper I, §6)

$$f_{\mathrm{crit}}(N) - b(N) = \frac{N^2}{8} - \frac{N^2}{12} + O(N) = \frac{N^2}{24} + O(N)$$

```python
import sys, math
sys.path.insert(0, '../src')
from planetary_polygons.extensions.equivariant_index import b_exact
from planetary_polygons.extensions.algebraic_thresholds import h2_stability_threshold
```

## 1. The exact offset $b(N)$

```python
print(f\"{'N':>4} {'b(N)':>12} {'N²/12':>10} {'ratio':>8}\")
for N in range(5, 20):
    b = b_exact(N)
    print(f'{N:4d} {b:12.6f} {N**2/12:10.4f} {b/(N**2/12):8.4f}')
```

## 2. Growth law: $f_{\mathrm{crit}} - b(N) \approx f_{\mathrm{crit}}/3$

```python
print(f\"{'N':>4} {'f_crit':>8} {'b(N)':>8} {'f-b':>8} {'(f-b)/f':>8}\")
for N in range(8, 20):
    xi, _, _ = h2_stability_threshold(N)
    if xi <= 0: continue
    m = N // 2
    f = m*(N-m)/2
    b = b_exact(N)
    residual = f - b
    print(f'{N:4d} {f:8.1f} {b:8.3f} {residual:8.3f} {residual/f:8.4f}')
print('\nRatio (f-b)/f → 1/3 as N → ∞')
```

## Summary

1. $b(N) = N(N{+}1)/12 - \log 2 + \log N/(N{-}1)$ verified.
2. Growth law: $f_{\mathrm{crit}} - b \to f_{\mathrm{crit}}/3$ with coefficient from $B_2 = 1/6$.
