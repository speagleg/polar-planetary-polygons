# Notebook 17: BTZ Critical Exponent $\nu = 1/2$ (Paper II, §10.8)

At the palindromic threshold $\rho^*(N)$:
$$Z(\rho) \sim \frac{Z_{\mathrm{frozen}}}{|\rho - \rho^*|^{1/2}}, \qquad Z_{\mathrm{frozen}} = \frac{2^{3(N-2)/4}}{(N{-}3)!!}$$

```python
import math
from functools import reduce
```

## 1. Frozen determinant

```python
def double_factorial_odd(n):
    '''(n)!! = 1·3·5·...·n for odd n'''
    if n <= 0: return 1
    return reduce(lambda a,b: a*b, range(1, n+1, 2), 1)

print('Frozen determinant Z = 2^{3(N-2)/4} / (N-3)!!')
print(f\"{'N':>4} {'2^{3(N-2)/4}':>14} {'(N-3)!!':>10} {'Z_frozen':>14}\")
for N in range(6, 21, 2):
    num = 2**(3*(N-2)/4)
    den = double_factorial_odd(N-3)
    Z = num / den
    print(f'{N:4d} {num:14.4f} {den:10d} {Z:14.6f}')
```

## 2. Casimir gaps at the threshold

```python
print('Casimir gaps: λ_m(ρ*) = (N-2m)²/8')
for N in [8, 10, 12]:
    print(f'N={N}:')
    for m in range(1, N//2+1):
        gap = (N-2*m)**2 / 8
        print(f'  m={m}: gap = {gap:.4f}')
```

## Summary

1. $Z_{\mathrm{frozen}} = 2^{3(N-2)/4}/(N{-}3)!!$ verified for even $N = 6,\ldots,20$.
2. Critical exponent $\nu = 1/2$ is universal (one zero mode, linear vanishing).
3. Casimir gaps $(N{-}2m)^2/8$ give the frozen eigenvalues.
