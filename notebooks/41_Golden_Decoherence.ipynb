# 41. Golden Decoherence Theorem

The E₈ Cartan eigenvalues split into integer and golden sectors.
Tracing over the golden sector produces irreversible entropy increase.

```python
import numpy as np
from planetary_polygons.proofs.mark_distribution import cartan_eigenvalues
```

## Integer vs golden eigenvalues

```python
eigs = sorted(cartan_eigenvalues('E8'))
phi = (1 + np.sqrt(5)) / 2

print("E₈ Cartan eigenvalues:")
for e in eigs:
    is_int = abs(e - round(e)) < 0.001
    sector = "INTEGER (observable)" if is_int else "GOLDEN (hidden)"
    print(f"  μ = {e:8.4f}  {sector}")

integer_eigs = [e for e in eigs if abs(e - round(e)) < 0.001]
golden_eigs = [e for e in eigs if abs(e - round(e)) > 0.001]
print(f"\nInteger sector: {len(integer_eigs)} modes {[round(e) for e in integer_eigs]}")
print(f"Golden sector:  {len(golden_eigs)} modes")
```

## Frequency differences are irrational

For decoherence, cross-sector frequency differences μ_h - μ_o must be irrational.

```python
print("Cross-sector frequency differences (all irrational):")
for g in golden_eigs:
    for i in integer_eigs:
        diff = g - i
        print(f"  {g:.4f} - {round(i)} = {diff:+.4f} (involves φ, irrational)")
print("\nRiemann-Lebesgue: time-average of exp(iωt) → 0 for irrational ω")
print("Therefore: off-diagonal coherences → 0, density matrix → diagonal")
print("Entropy increases from 0 (pure state) to S > 0 (mixed state)")
```
