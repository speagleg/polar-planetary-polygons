# Notebook 10: $N_{\mathrm{crit}}(\Delta)$ and RG Monotonicity (Paper II, §10.10)

The critical polygon number decreases with conformal dimension:
$$N_{\mathrm{crit}} = 7 \,(\Delta=0) \to 5 \,(\Delta=0.5) \to 4 \,(\Delta>2.4) \to 3 \,(\Delta\to\infty)$$

**Theorem (RG monotonicity).** The Casimir ratio $R(\Delta) = h(m_{\mathrm{crit}})/h(1)$ is strictly increasing, with $R(0) = m(N{-}m)/(N{-}1)$ and $R(\infty) = [\sin(\pi m/N)/\sin(\pi/N)]^2$.

```python
import sys, math
sys.path.insert(0, '../src')
from planetary_polygons.extensions.ncrit_delta import (
    ncrit_at_delta, stability_margin, find_critical_delta,
    generalized_casimir, casimir_ratio, weight_tilde
)
```

## 1. $N_{\mathrm{crit}}(\Delta)$ table

```python
print('N_crit(Δ) — flat-plane constrained Hessian')
print(f\"{'Δ':>6} {'N_crit':>8}\")
print('-' * 20)
for Delta in [0, 0.1, 0.5, 1.0, 2.0, 5.0, 10.0]:
    nc = ncrit_at_delta(Delta)
    print(f'{Delta:6.1f} {nc:8d}')
print('\nN_crit = 7 is unique to the logarithmic interaction.')
```

## 2. Critical $\Delta^*(N)$ values

```python
print('Critical Δ*(N):')
for N in [5, 6, 7]:
    if N == 7:
        print(f'  N={N}: Δ* = 0 (marginal)')
    else:
        D = find_critical_delta(N, 0.01, 5.0, tol=1e-8)
        print(f'  N={N}: Δ* = {D:.8f}')
```

## 3. Generalized Casimir $h(m, N, 0) = 2m(N{-}m)$

```python
print('Verify h(m,N,0) = 2m(N-m):')
for N in [5, 6, 7, 8]:
    for m in range(1, N):
        h = generalized_casimir(m, N, 0)
        expected = 2*m*(N-m)
        assert abs(h - expected) < 1e-8, f'FAIL: h({m},{N},0)={h}'
    print(f'  N={N}: all modes match ✓')
```

## 4. RG monotonicity: $R(\Delta)$ strictly increasing

```python
print('RG monotonicity verification:')
dDelta = 0.001
for N in range(4, 13):
    m = N // 2
    violations = 0
    for i in range(100):
        Delta = i * 0.1
        r1 = casimir_ratio(m, N, Delta)
        r2 = casimir_ratio(m, N, Delta + dDelta)
        if r2 < r1 - 1e-8:
            violations += 1
    print(f'  N={N:2d}: {"✓ MONOTONE" if violations==0 else f"✗ {violations} violations"}')

print('\nR(0) and R(∞) limits:')
for N in [5, 6, 7, 8]:
    m = N // 2
    R0 = m*(N-m)/(N-1)
    Rinf = (math.sin(math.pi*m/N)/math.sin(math.pi/N))**2
    R0_num = casimir_ratio(m, N, 0.0)
    Rinf_num = casimir_ratio(m, N, 50.0)
    print(f'  N={N}: R(0)={R0_num:.4f} (exact {R0:.4f}), R(∞)={Rinf_num:.4f} (exact {Rinf:.4f})')
```

## 5. Pentagon golden ratio: $\sin(2\pi/5)/\sin(\pi/5) = \varphi$

```python
phi = (1 + math.sqrt(5)) / 2
ratio = math.sin(2*math.pi/5) / math.sin(math.pi/5)
print(f'sin(2π/5)/sin(π/5) = {ratio:.15f}')
print(f'φ = (1+√5)/2       = {phi:.15f}')
print(f'Difference          = {abs(ratio-phi):.2e}')
print('\nThe pentagon identity is EXACT to machine precision.')
```
