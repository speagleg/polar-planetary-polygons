The hexagon ($N=6$) is **uniquely 2D**: stable at $d=2$, unstable at $d \ge 3$.

```python
import sys, math
import numpy as np
sys.path.insert(0, '../src')
from planetary_polygons.extensions.ncrit_delta import constrained_hessian_delta, stability_margin
```

## 1. $N_{\mathrm{crit}}$ by dimension

```python
print(f\"{'d':>4} {'s=d-2':>6} {'N':>4} {'margin':>10} {'stable':>8}\")
for d in [2, 3, 4, 5]:
    s = d - 2
    Delta = s / 2.0 if s > 0 else 0  # Δ = s/2 maps to the interaction
    # Actually use Δ = (d-2)/2 directly
    Delta = (d - 2) / 2
    for N in range(4, 10):
        m = stability_margin(N, Delta)
        stable = m >= -1e-4
        print(f'{d:4d} {Delta:6.2f} {N:4d} {m:+10.4f} {"✓" if stable else "✗":>8}')
```

## 2. Soft Havelock decomposition for $s = 3$

```python
def sigma_s(m, N, s):
    from math import pi, sin, cos
    return sum((1-cos(2*pi*p*m/N))/(2*sin(pi*p/N))**s for p in range(1, N))

print('Soft Havelock: σ_m ≈ a(N)·m(N-m)/2 + R_m')
for N in [6, 8, 10]:
    modes = list(range(1, N))
    sigma_vals = np.array([sigma_s(m, N, 3) for m in modes])
    casimir_vals = np.array([m*(N-m)/2 for m in modes])
    a = np.dot(sigma_vals, casimir_vals) / np.dot(casimir_vals, casimir_vals)
    R = sigma_vals - a * casimir_vals
    print(f'  N={N}: a = {a:.4f}, max|R| = {max(abs(R)):.4f}')
```

## Summary

1. $N_{\mathrm{crit}}(d=3) = 5$: the hexagon is a 2D phenomenon.
2. Soft Havelock decomposition preserves mode ordering for $d \ge 3$.
3. Only $s = 2$ (logarithmic) gives an exact polynomial Casimir.
