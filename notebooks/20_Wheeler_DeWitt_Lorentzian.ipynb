# Notebook 20: Wheeler-DeWitt Equation and Lorentzian Continuation (Paper II, \u00a710.8)

$$\left[-\frac{\hbar^2}{2c}\frac{d^2\Psi}{d\rho^2} + V(\rho)\right]\Psi = 0$$
where $V(\rho) = \log(2\sinh\rho) + b(N) - m(N{-}m)/2 + \delta_m$, $c = 12N^2$.

```python
import sys, math, numpy as np
sys.path.insert(0, '../src')
from planetary_polygons.extensions.wheeler_dewitt import (
    casimir, b_exact, wdw_potential, airy_solution,
    find_threshold
)
from planetary_polygons.extensions.lorentzian_continuation import C1_sinh, C1_cosh

def airy_width(N):
    """Airy quantum smoothing width: delta_rho = (24*N**2)**(-1/3)."""
    return (24 * N**2)**(-1/3)
```

## 1. WDW potential $V(\rho) = \lambda_{m^*}(\rho)$

```python
print("WDW potential V(\u03c1) = log(2sinh \u03c1) + b(N) - f(m*)")
print("Sign change at palindromic threshold \u03c1*\n")
for N in [8, 10, 12]:
    b = b_exact(N)
    m_star = N // 2
    f_m = casimir(m_star, N)
    rho_star = find_threshold(N)
    print(f"N = {N}, m* = {m_star}, f(m*) = {f_m:.2f}, b(N) = {b:.4f}, \u03c1* = {rho_star:.4f}")
    for rho in np.linspace(max(0.1, rho_star - 1), rho_star + 1, 7):
        V = wdw_potential(rho, N)
        sign = "+" if V > 0.001 else "-" if V < -0.001 else "\u22480"
        print(f"  \u03c1 = {rho:.3f}: V = {V:+.6f} [{sign}]")
    print()
```

## 2. Airy smoothing: $\delta\rho = (24N^2)^{-1/3}$

```python
print("Airy width \u03b4\u03c1 = 1/(24N\u00b2)^{1/3}")
print(f"{'N':>4} {'c=12N\u00b2':>10} {'\u03b4\u03c1':>14} {'\u03b4\u03c1\u00b7N^{2/3}':>14} {'\u21922^{-1/3}':>12}")
print("-" * 58)
target = 2**(-1/3)
for N in range(8, 21):
    dr = airy_width(N)
    ratio = dr * N**(2/3)
    print(f"{N:4d} {12*N**2:10d} {dr:14.6f} {ratio:14.6f} {target:12.6f}")
print(f"\nVerified: \u03b4\u03c1\u00b7N^{{2/3}} \u2192 2^{{-1/3}} = {target:.6f}")
```

## 3. Three regimes

```python
print("THREE REGIMES:\n")
print("1. CLASSICAL (c\u2192\u221e): V(\u03c1)=0 gives palindromic threshold = Einstein eq.")
print("2. SEMICLASSICAL (WKB): \u03a8 ~ exp(\u00b1(i/\u210f)\u222b\u221a(2cV)d\u03c1)")
print("3. QUANTUM (near threshold): \u03a8 ~ Ai((2c coth\u03c1*)^{1/3}(\u03c1-\u03c1*))\n")
for N in [8, 12, 20]:
    rho_star = find_threshold(N)
    width = airy_width(N)
    print(f"  N={N}: \u03c1* = {rho_star:.4f}, quantum width \u03b4\u03c1 = {width:.6f}")
```

## 4. Lorentzian continuation: sinh\u2194cosh

```python
print("sinh(\u03c1 + i\u03c0/2) = i\u00b7cosh(\u03c1)")
print("log(2sinh(\u03c1+i\u03c0/2)) = log(2cosh(\u03c1)) + i\u03c0/2\n")
print(f"{'\u03c1':>6} {'log(2sinh)':>16} {'log(2cosh)':>16} {'diff':>16}")
print("-" * 58)
for rho in [0.5, 1.0, 2.0, 3.0, 5.0, 10.0]:
    s = C1_sinh(rho)
    c = C1_cosh(rho)
    print(f"{rho:6.1f} {s:16.8f} {c:16.8f} {c-s:16.8f}")
print("\nDiff = log(coth \u03c1) \u2192 0 exponentially. Same IR, different UV.")
```

## 5. Three geometries from Im(\u03c1)

```python
print("| Im(\u03c1) | Geometry | Kernel         | K   |")
print("|---------|----------|----------------|-----|")
print("| 0       | H\u00b2       | log(2sinh \u03c1)   | -1  |")
print("| \u03c0/2     | R\u00b2       | log(2sin t)    |  0  |")
print("| \u03c0       | S\u00b2       | log(2cos t)    | +1  |")
print("\nWick rotation \u03c1 \u2192 it:")
for t in [0.3, 0.7, 1.0, 1.2, 1.5]:
    print(f"  t={t:.1f}: log(2sin t)={math.log(2*math.sin(t)):.6f}")
```

## 6. Universal resonance width \u0393 = \u03c0

```python
print("Universal resonance width \u0393 = \u03c0, lifetime \u03c4 = 1/\u03c0\n")
print(f"{'N':>4} {'\u03c1*':>10} {'\u0393':>10} {'\u03c4':>10}")
print("-" * 36)
for N in [6, 7, 8, 10, 12, 16, 20]:
    rho_star = find_threshold(N)
    if rho_star and rho_star > 0:
        print(f"{N:4d} {rho_star:10.4f} {math.pi:10.4f} {1/math.pi:10.4f}")
print(f"\n\u0393 = \u03c0 is UNIVERSAL (independent of N and m).")
```
