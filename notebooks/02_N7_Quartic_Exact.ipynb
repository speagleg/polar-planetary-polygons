# N=7 Quartic Stability: The Log-Scaling Identity

The regular heptagon ($N=7$) of identical point vortices sits at the **boundary** of
linear stability: the Havelock eigenvalue $\lambda_3 = 0$ creates a 2D neutral subspace.
Stability is then decided by the **quartic** term in the constrained energy expansion.

This notebook demonstrates the exact computation:

| Quantity | Exact value | Decimal |
|----------|-------------|----------|
| Unconstrained quartic | $153/7$ | $\approx 21.857$ |
| Log-scaling correction | $18/7$ | $\approx 2.571$ |
| **Constrained quartic** | **$135/7$** | $\approx 19.286$ |
| **$\alpha_0$** | **$45/14$** | $\approx 3.214$ |

The key insight is the **log-scaling identity**: the Newton projection onto the
angular-impulse surface introduces an additive correction that is exactly computable
in closed form.

```python
import sys
sys.path.insert(0, '../src')

import numpy as np
from fractions import Fraction
```

## 1. Unconstrained quartic = 153/7

The unconstrained quartic $d^4 H / d\varepsilon^4$ is evaluated on the angular
Fourier modes of the $N$-ring. For $N=7$, the neutral subspace corresponds to
$m=3$ (and its $\mathbb{Z}_7$-conjugate $m=4 = 7-3$).

By $\mathbb{Z}_7$ symmetry, the quartic is **constant** on the entire 2D neutral
subspace: both $m=3$ and $m=4$ give the same value.

```python
from planetary_polygons.extensions.n7_bifurcation import (
    unconstrained_quartic,
    quartic_exact_n7,
    constrained_quartic_exact,
    constraint_correction_exact,
    constrained_quartic_n7,
    alpha_0_exact,
)

# Compute unconstrained quartic for both neutral modes
q_m3 = unconstrained_quartic(7, 3)
q_m4 = unconstrained_quartic(7, 4)

print(f"Unconstrained quartic (m=3): {q_m3:.10f}")
print(f"Unconstrained quartic (m=4): {q_m4:.10f}")
print(f"Exact value 153/7:           {float(Fraction(153, 7)):.10f}")
print()
print(f"Z_7 symmetry check: |q(m=3) - q(m=4)| = {abs(q_m3 - q_m4):.2e}")
print(f"Agreement with 153/7:        |q(m=3) - 153/7| = {abs(q_m3 - float(Fraction(153, 7))):.2e}")

# Exact fraction
assert quartic_exact_n7() == Fraction(153, 7)
print(f"\nExact: quartic_exact_n7() = {quartic_exact_n7()} = {float(quartic_exact_n7()):.10f}  [CONFIRMED]")
```

## 2. The Log-Scaling Identity

The Newton projection onto the angular-impulse surface $\{L = N\}$ is:
$$
z(\varepsilon) = (z + \varepsilon u) \cdot \sqrt{\frac{N}{N + \varepsilon^2}}
$$

This scales all positions by $s(\varepsilon) = \sqrt{N/(N + \varepsilon^2)}$.
For the logarithmic Hamiltonian $H = -\sum_{j<k} \ln|z_j - z_k|$, each of the
$\binom{N}{2}$ pairs acquires an additive $-\ln s(\varepsilon)$, so:

$$
H_{\text{constrained}}(\varepsilon) = H_{\text{unconstrained}}(\varepsilon)
  + \frac{N(N-1)}{4} \ln\!\left(1 + \frac{\varepsilon^2}{N}\right)
$$

The 4th derivative of the correction at $\varepsilon = 0$:

$$
\ln\!\left(1 + \frac{\varepsilon^2}{N}\right) = \frac{\varepsilon^2}{N}
  - \frac{\varepsilon^4}{2N^2} + O(\varepsilon^6)
$$

$$
\frac{d^4}{d\varepsilon^4}\!\left[\frac{N(N-1)}{4} \ln\!\left(1 + \frac{\varepsilon^2}{N}\right)\right]_{\varepsilon=0}
= \frac{N(N-1)}{4} \cdot \frac{-12}{N^2} = -\frac{3(N-1)}{N}
$$

For $N=7$: correction $= -3 \cdot 6 / 7 = -18/7$.

Therefore: **constrained quartic** $= 153/7 - 18/7 = 135/7$.

```python
# Verify the log-scaling identity step by step
N = 7

# The correction: -3(N-1)/N
correction = Fraction(-3 * (N - 1), N)
print(f"Log-scaling correction (4th derivative): {correction} = {float(correction):.10f}")
print(f"Magnitude |correction| = {abs(correction)} = {float(abs(correction)):.10f}")
print()

# Constrained = unconstrained + correction
unconstrained = Fraction(153, 7)
constrained = unconstrained + correction
print(f"Unconstrained:  {unconstrained}")
print(f"Correction:    {correction}")
print(f"Constrained:    {unconstrained} + ({correction}) = {constrained}")
print()

# Verify against module functions
assert constrained_quartic_exact() == Fraction(135, 7)
print(f"constrained_quartic_exact() = {constrained_quartic_exact()}  [CONFIRMED]")

assert constraint_correction_exact() == Fraction(18, 7)
print(f"constraint_correction_exact() = {constraint_correction_exact()}  [CONFIRMED]")

# Verify the decomposition
assert quartic_exact_n7() - constraint_correction_exact() == constrained_quartic_exact()
print(f"\nDecomposition: 153/7 - 18/7 = {quartic_exact_n7()} - {constraint_correction_exact()} = {constrained_quartic_exact()}  [CONFIRMED]")
```

## 3. The Leading Amplitude Coefficient: $\alpha_0 = 45/14$

The constrained energy expansion on the neutral subspace is
$H(\varepsilon) = H(0) + \frac{1}{24} \cdot \frac{135}{7} \cdot \varepsilon^4 + O(\varepsilon^6)$,
so the leading amplitude coefficient is:

$$
\alpha_0 = \frac{1}{6} \cdot \frac{135}{7} = \frac{135}{42} = \frac{45}{14} \approx 3.214
$$

```python
alpha = alpha_0_exact()

print(f"alpha_0_exact() = {alpha}")
print(f"float(alpha_0) = {float(alpha):.10f}")
print()

assert alpha == Fraction(45, 14)
print(f"alpha_0 = 45/14  [CONFIRMED]")

# Verify the derivation: 135/7 / 6 = 135/42 = 45/14
assert Fraction(135, 7) / 6 == Fraction(45, 14)
print(f"135/7 / 6 = {Fraction(135, 7) / 6} = 45/14  [CONFIRMED]")
print()
print(f"Since alpha_0 = 45/14 > 0, the quartic is POSITIVE:")
print(f"the N=7 polygon is a constrained energy MINIMUM at quartic order.")
```

## 4. Numerical Verification via Richardson Extrapolation

As an independent check, we compute the constrained quartic numerically using
finite-difference Richardson extrapolation on the log-ratio $\Delta H$ method.
This should agree with the exact value $135/7 \approx 19.2857$ to high accuracy.

```python
q_numerical = constrained_quartic_n7()
q_exact = float(Fraction(135, 7))

print(f"Numerical (Richardson):  {q_numerical:.6f}")
print(f"Exact 135/7:             {q_exact:.6f}")
print(f"Difference:              {abs(q_numerical - q_exact):.6f}")
print()

agreement = abs(q_numerical - q_exact)
assert agreement < 0.005, f"Numerical does not agree with exact: diff = {agreement}"
print(f"|numerical - exact| = {agreement:.2e} < 0.005  [CONFIRMED]")
print()
print(f"The Richardson extrapolation independently confirms 135/7.")
```

## Summary

All exact results confirmed:

1. **Unconstrained quartic** $= 153/7$, verified numerically for both $m=3$ and $m=4$
   neutral modes ($\mathbb{Z}_7$ symmetry).

2. **Log-scaling identity**: the Newton projection onto $\{L=N\}$ introduces an
   additive correction whose 4th derivative is exactly $-3(N-1)/N = -18/7$ for $N=7$.

3. **Constrained quartic** $= 153/7 - 18/7 = 135/7$, independently confirmed by
   Richardson extrapolation to within $5 \times 10^{-3}$.

4. **Leading amplitude** $\alpha_0 = 45/14 \approx 3.214 > 0$, proving the $N=7$
   polygon is a **constrained energy minimum** at quartic order.

This is the sharpest stability result available for $N=7$: the quadratic term vanishes
($\lambda_3 = 0$), but the quartic term is strictly positive, establishing nonlinear
(Lyapunov) stability.
