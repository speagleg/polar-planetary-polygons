# Notebook 49: Instanton Proof for $V_{ub}$

**Paper IV, Theorem (thm:instanton-vub).** The $K^{N-1}$ suppression
of $V_{ub}$ from the texture zero $(YY^\dagger)_{02} = 0$ and the
winding number argument.

All results from `instanton_proof.py` (9 tests).

```python
from planetary_polygons.proofs.instanton_proof import (
    texture_zero_proof, winding_numbers, instanton_action, selectivity_proof,
)

# Texture zero
result = texture_zero_proof()
print(f"Texture: {result['texture']}")
print(f"Nonzero entries: {result['nonzero_count']}")
print(f"(YY^dag)_02 = 0? {result['YYdag_02_is_zero']}")
print(f"  Y_00 = 0: {result['Y00_zero']}")
print(f"  Y_21 = 0: {result['Y21_zero']}")
print(f"  Y_22 = 0: {result['Y22_zero']}")
```

## Winding Numbers

```python
table = winding_numbers()
print(f"{'Element':>8} {'Up modes':>12} {'Dn modes':>12} {'w_up':>5} {'w_dn':>5} {'w_tot':>5} {'Level':>10}")
print("-" * 65)
for r in table:
    print(f"{r['element']:>8} {r['up_modes']:>12} {r['dn_modes']:>12} {r['w_up']:>5} {r['w_dn']:>5} {r['w_total']:>5} {r['level']:>10}")
```

## Instanton Action and Selectivity

```python
action = instanton_action()
print(f"k_frac = {action['k_frac']:.6f}")
print(f"S_inst = (N-1) * 2*pi*k_frac = {action['N_minus_1']} * 2*pi*{action['k_frac']:.4f} = {action['S_inst']:.4f}")
print(f"exp(-S) = {action['exp_neg_S']:.6f}")
print(f"K^(N-1) = {action['K_to_N_minus_1']:.6f}")
print(f"Match: {abs(action['exp_neg_S'] - action['K_to_N_minus_1']) < 1e-10}")

sel = selectivity_proof()
print(f"\nSelectivity:")
print(f"  V_us corrected? {sel['V_us_corrected']} (should be False)")
print(f"  V_cb corrected? {sel['V_cb_corrected']} (should be False)")
print(f"  V_ub corrected? {sel['V_ub_corrected']} (should be True)")
print(f"  Reason: {sel['reason']}")
```
